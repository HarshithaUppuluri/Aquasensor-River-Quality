# 14 - Live Prediction Validation

This notebook investigates saved AquaPulse live forecasts and compares them with actual AquaSensor observations where corresponding future measurements are available.

```python
from pathlib import Path
import pandas as pd

PROJECT_ROOT = Path.cwd().parent

FORECAST_FILE = (
    PROJECT_ROOT
    / "data"
    / "processed"
    / "live_river_do_forecasts.csv"
)

HISTORY_FILE = (
    PROJECT_ROOT
    / "data"
    / "processed"
    / "live_aquasensor_api_readings.csv"
)

print("Forecast file exists:", FORECAST_FILE.exists())
print("History file exists:", HISTORY_FILE.exists())
```

```python
if FORECAST_FILE.exists():
    forecasts = pd.read_csv(
        FORECAST_FILE,
        low_memory=False
    )

    print("Forecast rows:", len(forecasts))
    display(forecasts.tail())
else:
    print("No saved live forecast file is currently available.")
```

```python
if HISTORY_FILE.exists():
    actual = pd.read_csv(
        HISTORY_FILE,
        low_memory=False
    )

    actual["timestamp"] = pd.to_datetime(
        actual["timestamp"],
        errors="coerce"
    )

    print("Actual live records:", len(actual))
    display(actual.tail())
else:
    print("No saved live AquaSensor history is currently available.")
```

## Validation Consideration

Live prediction validation depends on continuous collection of both forecasts and subsequent actual AquaSensor measurements.

As AquaPulse operates for longer periods, this dataset can be expanded to compare forecasted DO against the values that were later observed at the corresponding prediction timestamps.
