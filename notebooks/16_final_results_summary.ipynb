# 16 - Final AquaPulse Results Summary

## Project Overview

AquaPulse combines:

- AquaSensor river measurements
- Environmental and weather information
- Dissolved oxygen forecasting
- Linear Regression, Random Forest and XGBoost
- Eight prediction horizons from 15 to 120 minutes
- Anomaly detection
- SHAP explainability
- Live monitoring
- Interactive Flask visualisation

This notebook provides a final summary of the main computational outputs produced during the project.

```python
from pathlib import Path
import pandas as pd

PROJECT_ROOT = Path.cwd().parent

PERFORMANCE_FILE = (
    PROJECT_ROOT
    / "data"
    / "processed"
    / "do_prediction_model_performance.csv"
)

performance = pd.read_csv(
    PERFORMANCE_FILE
)

print("Total model evaluation records:", len(performance))
```

```python
do_results = performance[
    performance["target_type"]
    .astype(str)
    .str.lower()
    == "do mg/l"
].copy()

best_results = (
    do_results
    .sort_values(
        ["horizon", "R2"],
        ascending=[True, False]
    )
    .groupby("horizon", as_index=False)
    .first()
)

best_results[
    [
        "horizon",
        "model",
        "MAE",
        "RMSE",
        "R2"
    ]
]
```

```python
print("Best model by forecast horizon:")

for _, row in best_results.iterrows():

    print(
        f"{row['horizon']}: "
        f"{row['model']} | "
        f"R²={row['R2']:.4f} | "
        f"RMSE={row['RMSE']:.4f}"
    )
```
