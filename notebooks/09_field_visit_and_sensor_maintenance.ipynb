## Purpose

This notebook documents practical observations associated with the AquaSensor monitoring system used for the River Derwent project.

The purpose is to provide supporting evidence that live environmental monitoring depends not only on software and machine-learning models, but also on the continued operation and maintenance of physical IoT sensors.

## Field Visit - 27 July 2026

A field-related meeting took place on 27 July 2026 concerning the AquaSensor monitoring system.

During this period, a sensor battery had become discharged and the sensor could no longer continue normal monitoring. Removed the affected equipment so that the battery could be charged. After charging, the equipment was placed back into operation.

This observation was important to the AquaPulse project because it demonstrated a practical limitation of live IoT environmental monitoring. Interruptions in power or sensor maintenance can produce gaps in live data even when the software and machine-learning components are functioning correctly.

The event therefore helped explain why missing or irregular periods may appear within AquaSensor records and why the AquaPulse system should distinguish between sensor-data availability and application/model performance.

## Impact on AquaPulse

This observation influenced the interpretation of live data in AquaPulse. A missing reading does not automatically indicate a software failure, prediction failure or environmental anomaly. The absence of data may instead be caused by sensor maintenance, battery depletion, communication problems or temporary API availability.

For this reason, AquaPulse retains previously collected readings and includes fallback behaviour when new AquaSensor information is temporarily unavailable.
